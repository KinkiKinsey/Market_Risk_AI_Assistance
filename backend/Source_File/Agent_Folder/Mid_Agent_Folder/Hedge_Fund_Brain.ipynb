````python

import os
import json
from typing import TypedDict, Optional, Dict
from langchain_deepseek import ChatDeepSeek
from langgraph.graph import StateGraph, START, END


os.environ["DEEPSEEK_API_KEY"] = "sk-43e9043c7ab8480393d34367f2ae997e"

class EarningsPipelineState(TypedDict):
    future_development: str
    transcript: str
    qa_pairs: Optional[Dict[str, str]]
    buy_side_focus: Optional[str]

extractor_llm = ChatDeepSeek(
    model="deepseek-chat",       # ✅ FIXED: Use chat model instead of reasoner
    temperature=0.5,             # more flexible for inference
    max_tokens=2000             # ✅ FIXED: Increased tokens for longer responses
)

analyst_llm = ChatDeepSeek(
    model="deepseek-chat",       # summarization model for buy-side focus
    temperature=0.5,
    max_tokens=800              # ✅ INCREASED: More tokens for 4-bullet analysis
)

def extract_qa_pairs(state: EarningsPipelineState) -> EarningsPipelineState:
    """
    Extracts or infers Q&A pairs from transcript and future guidance.
    Retries automatically with interpretive mode if none detected.
    """
    base_prompt = f"""
    You are a professional sell-side analyst parsing an earnings call.

    Read the transcript and infer QUESTION / ANSWER pairs between analysts and executives
    — even if they are not explicitly labeled.

    Transcript:
    {state['transcript']}

    Your goal:
    - Identify 2–5 meaningful Q&A pairs about performance, strategy, guidance, or risks.
    - Use real questions if present; otherwise infer likely ones from management discussion.
    - Return ONLY valid JSON in this format:
      {{
        "Q1": "What are your growth priorities?",
        "A1": "Target focuses on digital expansion and AI-driven efficiency.",
        "Q2": "How will margins evolve?",
        "A2": "Cost controls and automation are improving profitability."
      }}
    """

    response = extractor_llm.invoke(base_prompt)
    text = response.content.strip()

    # --- 🧹 Clean JSON fences if DeepSeek returns markdown
    if "```json" in text:
        text = text.split("```json")[-1].split("```")[0].strip()

    # --- 🔁 Retry if empty output
    if not text:
        print("⚠️ [Warning] Empty extraction. Retrying with fallback reasoning...")
        retry_prompt = f"""
        No explicit questions detected earlier.
        Infer the all investor concerns (Q1, Q2, Q3...) and give short management answers (A1, A2, A3...)
        based on the transcript tone and strategic priorities.

        Transcript:
        {state['transcript']}

        Future Development:
        {state['future_development']}

        Return ONLY valid JSON:
        {{
            "Q1": "...", "A1": "...",
            "Q2": "...", "A2": "...",
            "Q3": "...", "A3": "..."
        }}
        """
        response = extractor_llm.invoke(retry_prompt)
        text = response.content.strip()

    # --- ✅ Final JSON cleaning
    if "```json" in text:
        text = text.split("```json")[-1].split("```")[0].strip()

    # --- 🧠 Try to parse JSON
    try:
        qa_json = json.loads(text)
        # If successful, check if it has Q&A pairs
        valid_qs = len([k for k in qa_json.keys() if k.startswith("Q")])
        print(f"✅ Extracted or inferred {valid_qs} question(s).")
    except json.JSONDecodeError:
        # ✅ FIXED: Parse incomplete/malformed JSON into readable Q&A format
        print("⚠️ JSON parsing failed. Extracting Q&A from text...")
        qa_json = {}
        
        # Try to extract Q&A pairs using regex
        import re
        # Match patterns like "Q1": "...", "A1": "..."
        q_pattern = r'"Q(\d+)":\s*"([^"]*)"'
        a_pattern = r'"A(\d+)":\s*"([^"]*)"'
        
        questions = re.findall(q_pattern, text)
        answers = re.findall(a_pattern, text)
        
        # Build clean Q&A dictionary
        for q_num, q_text in questions:
            qa_json[f"Q{q_num}"] = q_text
        for a_num, a_text in answers:
            qa_json[f"A{a_num}"] = a_text
        
        # If regex extraction failed, create a readable summary
        if not qa_json:
            print("⚠️ Using fallback: Converting raw text to readable format")
            qa_json = {
                "Summary": "Unable to parse structured Q&A. Key points from transcript:",
                "Content": text[:500] + "..." if len(text) > 500 else text
            }
        
        valid_qs = len([k for k in qa_json.keys() if k.startswith("Q")])
        print(f"✅ Recovered {valid_qs} question(s) from malformed JSON.")

    state["qa_pairs"] = qa_json
    return state

def generate_buy_side_focus(state: EarningsPipelineState) -> EarningsPipelineState:
    """
    ✅ ENHANCED: Combines future development strategy + Q&A to identify 
    what buy-side investors should focus on regarding the company's strategic direction.
    """
    prompt = f"""
    You are a seasoned buy-side equity analyst conducting deep dive analysis.

    COMPANY'S FUTURE DEVELOPMENT STRATEGY:
    {state['future_development']}

    EARNINGS CALL Q&A SUMMARY:
    {json.dumps(state['qa_pairs'], indent=2)}


    Future Development (This is what the company is actually focus on to overcome/mantain its current situation):
    {state['future_development']}

    TASK:
    Analyze BOTH the company's stated strategic direction AND the Q&A insights together.
    
    Identify what buy-side investors should focus on by answering:
    1. What are the company's all strategic priorities to overcome challenges OR maintain success?
    
    Focus points:
    - In term how rev growth 
    - In term how to improve Sale
    - In term how Executive Work (all dimension)

    return only 3 thing is fine 

    Keep each bullet ≤2 lines. Focus on ACTIONABLE insights.

    Format:
    [Strategic Priority + Action]

    """

    response = analyst_llm.invoke(prompt)
    text = response.content.strip()

    # --- Formatting fix
    if not text.startswith("1."):
        text = "1. " + text

    state["buy_side_focus"] = text
    return state

graph = StateGraph(EarningsPipelineState)
graph.add_node("extract_qa", extract_qa_pairs)
graph.add_node("buy_side_focus", generate_buy_side_focus)

graph.add_edge(START, "extract_qa")
graph.add_edge("extract_qa", "buy_side_focus")
graph.add_edge("buy_side_focus", END)

earnings_pipeline = graph.compile()


future_dev = future_and_earnings_result.future_development
transcript = future_and_earnings_result.transcript

input_state = {
    "future_development": future_dev,
    "transcript": transcript,
    "qa_pairs": None,
    "buy_side_focus": None,
}


result = earnings_pipeline.invoke(input_state)


print("\n" + "="*70)
print("🧾 EARNINGS CALL Q&A ANALYSIS")
print("="*70)

qa_data = result["qa_pairs"]
````

## Brain Structure

````python
# ==============================================================
# 1. Imports & Setup
# ==============================================================

import os, json
from datetime import datetime
from typing import TypedDict, List, Dict, Any
from langgraph.graph import StateGraph, START, END
from langchain_deepseek import ChatDeepSeek

# ==============================================================
# 2. Initialize LLM Tool
# ==============================================================

os.environ["DEEPSEEK_API_KEY"] = "sk-43e9043c7ab8480393d34367f2ae997e"

llm = ChatDeepSeek(
    model="deepseek-chat",
    temperature=0.3,
    max_tokens=4000
)

def llm_tool(prompt: str) -> str:
    """Simple LLM wrapper"""
    response = llm.invoke(prompt)
    return response.content


# ==============================================================
# 3. Brain State Definition (FIXED)
# ==============================================================

class BrainState(TypedDict):
    Macro: List[Dict[str, Any]]
    Sector: List[Dict[str, Any]]
    Market: List[Dict[str, Any]]
    Micro: List[Dict[str, Any]]
    inputs_to_process: List[Dict[str, str]]  # FIXED: Store inputs here
    buy_side_focus: str  # FIXED: Store buy-side focus here
    alpha_insights: Dict[str, Any]
    last_updated: str


# ==============================================================
# 4. LangGraph Node Definitions (FIXED)
# ==============================================================

# ---------- (1) Extractor Node (FIXED) ----------
def extractor_node(state: BrainState) -> BrainState:
    """Extract features from all inputs"""
    inputs_to_process = state.get("inputs_to_process", [])
    
    for input_item in inputs_to_process:
        text = input_item.get("text", "")
        context_name = input_item.get("context_name", "Unknown")
        
        if not text or len(str(text).strip()) == 0:
            continue

        prompt = f"""
You are a financial research extractor.
Identify which analytical layer (Macro, Sector, Market, Micro) this text belongs to,
extract the main factor, and list 2–5 concrete features (unique truths or structural insights).

Output JSON ONLY in this format:
{{
    "layer": "Macro|Sector|Market|Micro",
    "factor": "...",
    "features": [
        {{"statement": "...", "evidence": "...", "confidence": 0.85}}
    ]
}}

Context: {context_name}
Text:
{str(text)[:1500]}...
"""

        try:
            result_text = llm_tool(prompt)
            if "```json" in result_text:
                result_text = result_text.split("```json")[-1].split("```")[0].strip()
            result = json.loads(result_text)
        except Exception as e:
            print(f"⚠️ Extraction failed for {context_name}: {e}")
            continue

        layer = result.get("layer")
        if layer in ["Macro", "Sector", "Market", "Micro"]:
            factor_node = {
                "factor": result["factor"],
                "features": result["features"],
                "context": context_name,
                "created_at": str(datetime.now())
            }
            state[layer].append(factor_node)
            print(f"✅ {layer}: +{len(result['features'])} features from {context_name}")

    return state


# ---------- (2) Reflection Node (FIXED) ----------
def reflect_node(state: BrainState) -> BrainState:
    """Merge duplicates & rewrite unclear reasoning."""
    for layer in ["Macro", "Sector", "Market", "Micro"]:
        if not state.get(layer):
            continue

        prompt = f"""
You are a reflection agent.
Review the following {layer} layer data, merge duplicates,
clarify vague reasoning, and keep only top 16 by confidence.

Input:
{json.dumps(state[layer], indent=2)}

Output JSON list only (same structure).
"""

        try:
            result_text = llm_tool(prompt)
            if "```json" in result_text:
                result_text = result_text.split("```json")[-1].split("```")[0].strip()
            refined = json.loads(result_text)
        except Exception as e:
            print(f"⚠️ Reflection failed for {layer}: {e}")
            refined = state[layer]

        state[layer] = refined[:16]
        print(f"✅ Reflected {layer}: {len(state[layer])} factors")

    state["last_updated"] = str(datetime.now())
    return state


def alpha_node(state: BrainState) -> BrainState:
    """Discover hidden alpha insights"""
    buy_side_focus = state.get("buy_side_focus", "")
    
    # ✅ FIX: Build dict OUTSIDE the f-string
    brain_data = {
        "Macro": state.get("Macro", []),
        "Sector": state.get("Sector", []),
        "Market": state.get("Market", []),
        "Micro": state.get("Micro", [])
    }
    
    prompt = f"""
You are an Alpha Discovery Analyst.

Use the ticker's brain below:
{json.dumps(brain_data, indent=2)}

Buy-side current focus:
{buy_side_focus}

Task:
- Identify which features are *already priced in*.
- Identify which ones are *under-discussed* (hidden alpha).
- Return top 3 alpha features with reasons.

Output JSON:
{{
  "alpha_insights": [
    {{
      "feature": "...",
      "layer": "...",
      "reason_alpha": "...",
      "visibility": "Low",
      "expected_impact": "High",
      "confidence": 0.85
    }}
  ]
}}
"""

    try:
        result_text = llm_tool(prompt)
        if "```json" in result_text:
            result_text = result_text.split("```json")[-1].split("```")[0].strip()
        alpha = json.loads(result_text)
        print(f"✅ Discovered {len(alpha.get('alpha_insights', []))} alpha insights")
    except Exception as e:
        print(f"⚠️ Alpha discovery failed: {e}")
        alpha = {"alpha_insights": []}

    state["alpha_insights"] = alpha
    return state

# ==============================================================
# 5. Build & Compile LangGraph (FIXED)
# ==============================================================

graph = StateGraph(BrainState)
graph.add_node("extractor", extractor_node)
graph.add_node("reflector", reflect_node)
graph.add_node("alpha", alpha_node)

graph.add_edge(START, "extractor")
graph.add_edge("extractor", "reflector")
graph.add_edge("reflector", "alpha")
graph.add_edge("alpha", END)

app = graph.compile()

print("✅ Brain workflow compiled!")

# ==============================================================
# 6. Prepare Input Data (FIXED)
# ==============================================================

# Define your input variables (update these with your actual data)
inputs_list = [
    {"text": str(macro_result.analysis) if hasattr(macro_result, 'analysis') else str(macro_result), 
     "context_name": "Macro environment"},
    
    {"text": sector_result.get_answer("sector_trend"), 
     "context_name": "Industry trend"},
    
    {"text": sector_result.get_answer("company_competitor_landscape"), 
     "context_name": "Competitive landscape"},
    
    {"text": sector_result.competitor_summary, 
     "context_name": "Competitor summary"},
    
    {"text": future_and_earnings_result.future_development, 
     "context_name": "Company future plan"},
    
    {"text": str(revenue_result.business_segments), 
     "context_name": "Revenue breakdown"},
    
    {"text": str(revenue_result.cost_segments), 
     "context_name": "Cost structure"},
    
    {"text": str(revenue_result.supplier_segments), 
     "context_name": "Supplier analysis"},
]

# Buy-side focus from earnings pipeline
buy_side_focus_text = result.get("buy_side_focus", "Focus on growth and profitability") if 'result' in locals() else "Focus on growth and profitability"

print(f"✅ Prepared {len(inputs_list)} inputs")

# ==============================================================
# 7. Run the Workflow (FIXED)
# ==============================================================

# Initialize state with all inputs
initial_state = {
    "Macro": [],
    "Sector": [],
    "Market": [],
    "Micro": [],
    "inputs_to_process": inputs_list,
    "buy_side_focus": buy_side_focus_text,
    "alpha_insights": {},
    "last_updated": str(datetime.now())
}

print("🚀 Running Brain Workflow...")
print("="*70)

# Run the complete graph (one call does everything)
final_state = app.invoke(initial_state)

print("\n" + "="*70)
print("✅ Workflow Complete!")
print("="*70)

# ==============================================================
# 8. Display Results (FIXED)
# ==============================================================

print("\n🧠 === BRAIN SUMMARY ===")
print("="*70)
print(f"Macro Factors:  {len(final_state['Macro'])}")
print(f"Sector Factors: {len(final_state['Sector'])}")
print(f"Market Factors: {len(final_state['Market'])}")
print(f"Micro Factors:  {len(final_state['Micro'])}")

print("\n💡 === ALPHA INSIGHTS ===")
print("="*70)
alpha_data = final_state.get("alpha_insights", {})
insights = alpha_data.get("alpha_insights", [])

if insights:
    for i, insight in enumerate(insights, 1):
        print(f"\n{i}. 🎯 {insight.get('feature', 'N/A')}")
        print(f"   Layer: {insight.get('layer', 'N/A')}")
        print(f"   Visibility: {insight.get('visibility', 'N/A')}")
        print(f"   Impact: {insight.get('expected_impact', 'N/A')}")
        print(f"   Confidence: {insight.get('confidence', 0):.2f}")
        print(f"   Reason: {insight.get('reason_alpha', 'N/A')}")
else:
    print("⚠️ No alpha insights discovered")

print("\n" + "="*70)
print(f"✅ Analysis complete at {final_state['last_updated']}")
````
